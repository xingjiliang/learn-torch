```python
import torch
from statsmodels.stats.libqsturng import v_keys
from torch import nn
```

```python
class LayerNorm(nn.Module):
    def __init__(self, num_dims, epsilon=0e-5):
        super().__init__()
        self.alpha = torch.ones(num_dims, requires_grad=True).reshape(1, 1, num_dims)
        self.beta = torch.zeros(num_dims, requires_grad=True).reshape(1, 1, num_dims)
        self.epsilon = epsilon

    def forward(self, x):
        x_mean = x.mean(-1, keepdim=True)  #shape=(batch_size, seq, 1)
        #虽然可以直接求标准差...
        x_var = x.var(-1, keepdim=True)
        return (x - x_mean) / torch.sqrt(x_var + self.epsilon) * self.alpha + self.beta
```

```python
def softmax(x, dim=-1):
    x_exp = x.exp()
    x_exp_sum = x_exp.sum(dim=dim, keepdim=True)
    return x_exp / x_exp_sum
```

```python
class EncoderLayer(nn.Module):
    def __init__(self, num_dims, num_heads, bias=False, **kwargs):
        """
        @Author: XingJiliang
        :param num_heads: 
        :param num_dims: 在这里等于词嵌入的维度，设计为可被num_heads整除
        """
        super().__init__()
        self.W_q = nn.LazyLinear(num_dims, bias=bias)
        self.W_k = nn.LazyLinear(num_dims, bias=bias)
        self.W_v = nn.LazyLinear(num_dims, bias=bias)
        self.ln = LayerNorm(num_dims)
        
    def forward(self, x):
        batch_size = x.shape(0)
        seq_len = x.shape(1)
        q = self.W_q(x)
        k = self.W_k(x)
        v = self.W_v(x)
        q_slice = q.reshape(batch_size, seq_len, self.num_heads, -1).permute(0,2,1,3)
        k_slice = k.reshape(batch_size, seq_len, self.num_heads, -1).permute(0,2,3,1)
        v_slice = v.reshape(batch_size, seq_len, self.num_heads, -1).permute(0,2,1,3)
        
        attention_score = q_slice.matmul(k_slice)
        attention_weights = softmax(attention_score)
        y = torch.bmm(attention_weights, v_slice).permute(0,2,1,3).reshape(batch_size, seq_len, -1) #shape=[bch, heads, seq, dim]
        return x + LayerNorm(y) 
```

```python
class Encoder(nn.Module):
    def __init__(self, num_layers, num_dims, num_heads, **kwargs):
        super().__init__()
        self.encoder_layers = nn.Sequential()
        for i in range(num_layers):
            self.encoder_layers.add_module("encoder_layer" + i, EncoderLayer(num_dims, num_heads))

    def forward(self, x):
        return self.encoder_layers(x)
```
